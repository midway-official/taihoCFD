# Taiho-CFD 网格生成

网格生成逻辑位于 `tools/generate_mesh.py`，这里仅保留一个可重复执行的 notebook 入口。
数组行从 `y=0` 到 `y=Ly` 排列，因此最后一行是顶盖；这与当前 C++ 求解器的几何和边界索引一致。

````python
from pathlib import Path
import sys

TOOLS = Path.cwd() / 'tools'
if str(TOOLS) not in sys.path:
    sys.path.insert(0, str(TOOLS))

from generate_mesh import generate_lid_driven_cavity, generate_poiseuille_flow

# 顶盖驱动方腔：zoneid[-1, :] 是顶盖速度区。
generate_lid_driven_cavity(
    nx=64, ny=64, Lx=1.0, Ly=1.0,
    lid_u=1.0, lid_v=0.0,
    alpha_x=4.0, alpha_y=4.0,
    output_dir='examples/meshes/ldc_stretched',
)

# 泊肃叶流：左侧速度入口，右侧 p=0 压力出口。
generate_poiseuille_flow(
    nx=200, ny=80, Lx=3.0, Ly=0.2,
    inlet_u=1.0, inlet_v=0.0,
    alpha_y=4.0,
    output_dir='examples/meshes/poiseuille',
)
````

命令行等价用法：
```bash
python3 tools/generate_mesh.py cavity --nx 64 --ny 64 --alpha-x 4 --alpha-y 4 --output-dir examples/meshes/ldc_stretched
python3 tools/generate_mesh.py poiseuille --nx 200 --ny 80 --lx 3 --ly 0.2 --alpha-y 4 --output-dir examples/meshes/poiseuille
```
