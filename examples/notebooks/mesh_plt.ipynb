# 网格检查

当前坐标约定是 `x` 向右、`y` 向上；不再对 y 轴反转。

```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

mesh_dir = Path('examples/meshes/ldc_stretched')
X = np.loadtxt(mesh_dir / 'x.dat')
Y = np.loadtxt(mesh_dir / 'y.dat')
ny, nx = X.shape[0] - 1, X.shape[1] - 1

fig, ax = plt.subplots(figsize=(7, 7))
for i in range(X.shape[0]):
    ax.plot(X[i, :], Y[i, :], color='steelblue', linewidth=0.5)
for j in range(X.shape[1]):
    ax.plot(X[:, j], Y[:, j], color='steelblue', linewidth=0.5)
ax.set(title=f'Mesh ({ny} x {nx} cells)', xlabel='x', ylabel='y', aspect='equal')
fig.tight_layout()
plt.show()
```
